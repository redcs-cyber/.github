```python
import subprocess
import sys

print("Attempting to install compatible tokenizers and transformers versions...")

# First, install a compatible tokenizers version with wheels for Python 3.12
print("Installing tokenizers==0.15.2...")
process_tokenizers = subprocess.run([sys.executable, "-m", "pip", "install", "tokenizers==0.15.2", "--force-reinstall"], capture_output=True, text=True)
print("pip tokenizers stdout:", process_tokenizers.stdout)
print("pip tokenizers stderr:", process_tokenizers.stderr)
if process_tokenizers.returncode != 0:
    print(f"Error during tokenizers installation: {process_tokenizers.returncode}")
    print("Please check the error messages above. A manual fix might be needed for tokenizers.")
else:
    print("tokenizers installed successfully.")

# Then, install a compatible transformers version
print("Installing transformers==4.35.0...")
process_transformers = subprocess.run([sys.executable, "-m", "pip", "install", "transformers==4.35.0", "--force-reinstall"], capture_output=True, text=True)
print("pip transformers stdout:", process_transformers.stdout)
print("pip transformers stderr:", process_transformers.stderr)
if process_transformers.returncode != 0:
    print(f"Error during transformers installation: {process_transformers.returncode}")
    print("Please check the error messages above. A manual fix might be needed for transformers.")
else:
    print("transformers installed successfully.")

# Verify the installed versions
print("Verifying transformers version...")
process_version_transformers = subprocess.run([sys.executable, "-m", "pip", "show", "transformers"], capture_output=True, text=True)
print("pip show transformers output:\n", process_version_transformers.stdout)

print("Verifying tokenizers version...")
process_version_tokenizers = subprocess.run([sys.executable, "-m", "pip", "show", "tokenizers"], capture_output=True, text=True)
print("pip show tokenizers output:\n", process_version_tokenizers.stdout)
```
